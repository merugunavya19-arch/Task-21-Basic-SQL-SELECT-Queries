```python

import sqlite3
import pandas as pd

# Create an in-memory database
conn = sqlite3.connect(":memory:")
cursor = conn.cursor()

# Create a Customers table
cursor.execute("""
CREATE TABLE Customers (
    CustomerID INTEGER,
    CustomerName TEXT,
    City TEXT
)
""")

# Insert sample customer data
customers = [
    (1, 'Rahul', 'Hyderabad'),
    (2, 'Priya', 'Vijayawada'),
    (3, 'Arjun', 'Guntur'),
    (4, 'Sneha', 'Hyderabad'),
    (5, 'Kiran', 'Chennai')
]

cursor.executemany(
    "INSERT INTO Customers VALUES (?, ?, ?)",
    customers
)
conn.commit()

print("Customers table created successfully!")
```

```python

query = """
SELECT CustomerName, City
FROM Customers;
"""

result = pd.read_sql_query(query, conn)
display(result)
```
